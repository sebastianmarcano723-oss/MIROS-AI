```python
%%writefile app.py
import streamlit as st
from PIL import Image
import pandas as pd

# Configuración estética de la App
st.set_page_config(page_title='IA Multi-Tarea', layout='wide')

# --- Sidebar (Menú Lateral personalizado) ---
try:
    logo = Image.open('/content/WhatsApp Image 2026-07-26 at 1.01.45 PM.jpeg')
    st.sidebar.image(logo, use_container_width=True)
except:
    st.sidebar.title('LOGO APP')

st.sidebar.markdown('### Panel de Control')
menu = st.sidebar.radio(
    'Selecciona una función:',
    ['Generar Imágenes',
     'Presentaciones en PDF',
     'Organizar Calendario',
     'Subir Información (RAG)',
     'Búsqueda Web']
)

# --- Referencia Visual (Segunda Foto) ---
# Este bloque simula el apartado visual solicitado basado en la segunda imagen
st.markdown('---')

if menu == 'Generar Imágenes':
    st.header('🎨 Generación de Imágenes')
    prompt = st.text_area('Describe lo que quieres ver:')
    if st.button('Generar Arte'):
        st.info('Conectando con el motor de IA para generar tu imagen...')

elif menu == 'Presentaciones en PDF':
    st.header('📊 Creador de Presentaciones')
    tema = st.text_input('Tema de la presentación:')
    if st.button('Generar Estructura y PDF'):
        st.success('Documento generado con lógica contextual.')

elif menu == 'Organizar Calendario':
    st.header('📅 Calendario de Evaluaciones')
    uploaded_plan = st.file_uploader('Sube tu Plan de Evaluación', type=['pdf', 'jpg', 'txt'])
    if uploaded_plan:
        st.write('IA analizando fechas y temas pendientes...')
        # Ejemplo de organización lógica
        df = pd.DataFrame({'Tarea': ['Estudio Parcial', 'Entrega Proyecto'], 'Fecha': ['2024-06-15', '2024-06-20']})
        st.table(df)

elif menu == 'Subir Información (RAG)':
    st.header('📁 Memoria Personalizada')
    st.info('Sube archivos (PDF, Audio, Drive, YT) para que la IA aprenda de ellos.')
    files = st.file_uploader('Archivos:', accept_multiple_files=True)
    url = st.text_input('URL (Sitio Web o YouTube):')
    if st.button('Entrenar con mis datos'):
        st.success('La IA ahora responderá basada únicamente en tu contenido.')

elif menu == 'Búsqueda Web':
    st.header('🔍 Búsqueda Inteligente')
    query = st.text_input('¿Qué deseas investigar en la red?')
    if st.button('Buscar Fuentes'):
        st.write('Consultando fuentes confiables y sintetizando respuesta...')
```
